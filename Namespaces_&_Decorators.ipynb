## Namespaces
- A Namespace is a space that holds names(identifiers). Programattically speaking, namespaces are dictionary of identifiers(keys) and thier objective(values).
There are 4 types of namespaces.
- Builtin Namespaces
- Global Namespaces
- Enclosing Namespaces
- Local Namespaces


## Scope and LEGB Rule
A scope is a texual region of a Python Programm where a namespace is directly accessible.
The interpretor searches for the name from inside out. looking in the local.enclosing,global,and finally the builtin scope. if the interpretor does'nt find the name in any of these locations,then python raises a NameError exception.


```python
# local and global


a = 2   # global Var
def temp():#local var
  b=3
  print(b)
temp()
print(a)
```

```python
# local and global -->> same name
a=2
def temp():
  #local var
  a=3
  print(a)
temp()
print(a)
```

```python
# local and global -> local does not have but  global has
a = 2
def temp():
  #local
  print(a)

temp()
print(a)
```

```python
a =2
def temp():
  #local var
  global a
  a+=1
  print(a)

temp()
print(a)
```

```python
# global created inside local
def temp():
  #local var
  global a
  a=1
  print(a)

temp()
print(a)
```

```python
# local and global --> function paramter is local
def temp(x):
  print(x)

a=3
print(a)
temp(4)
print(x)
```

```python
# LEGB RULE
# Enclosing scope
def outer():
  def inner():
    print(a)
  inner()
  print("outer function")

outer()
print("main programm")
```

```python
# non local keyword
def outer():
  a = 1
  def inner():
    nonlocal a
    a+=1
    print("inner",a)
  inner()
  print("outer",a)

outer()
print("main programm")

```

# Decorators
A decorators in python is a function that receives another function as input an input and adds some functionalities to it and returns it.<br>
This can happen only because python functions are 1st class citizens.<br>
There are 2 types of decorators available in python.<br>

- `Built in decorators` like `@staticmethod`,`@classmethod`,`@abstractmethod`,and @property etc.<br>




```python
# Python are 1st class citizens
def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
```

```python
#simple example
def my_decorator(func):
  def wrapper():
    print("***************************")
    func()
    print("***************************")
  return wrapper

def hello():
  print("hello")

def display():
  print("hello nitish")

a=my_decorator(hello)
a()

b=my_decorator(display)
b()
```
